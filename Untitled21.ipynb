```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python

plt.figure(figsize=(7,4))
sns.barplot(data=df,x="Product line", y="Total",estimator="sum")
plt.title("Total Sales by Product line")
plt.xlabel("Product line")
plt.ylabel("Total sales")
plt.xticks(rotation=45)
plt.legend(["Total sales"])
```

```python
daily_sales=df.groupby("Date")["Total"].sum()
plt.figure(figsize=(7,4))
plt.plot(daily_sales.index,daily_sales.values,marker="o")
plt.title("Daily Total Sales")
plt.xlabel("Date")
plt.ylabel("Toale sales")
plt.legend(["Total sales"])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df,x="Unit price",bins=10)
plt.title("Distribution of Unit Price")
plt.xlabel("Unit price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df,x="Unit price", y="Quantity")
plt.title("Unit Price vs Quantity")
plt.xlabel("Unit price")
plt.ylabel("Quantity")
plt.tight_layout()
plt.show()
```

```python
payment_counts=df["Payment"].value_counts()
plt.figure(figsize=(4,4))
plt.pie(payment_counts,labels=payment_counts.index,autopct="%1.1f%%")
plt.title("payment method distribution")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,y="Total")
plt.title("Distribution of total sales")
plt.xlabel("Sales")
plt.ylabel("Total sales")
plt.tight_layout()
           
plt.show()
```

```python
plt.figure(figsize=(8,5))
correlation=df.corr(numeric_only=True)
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.barplot(data=df,x="City",y="Total",estimator="sum")
plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
df.columns
```

```python
plt.figure(figsize=(5,4))
sns.countplot(data=df,x="Customer type")
plt.title("Customer Type Distribution")
plt.xlabel("Customer type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.violinplot(data=df,x="Gender",y="Rating")
plt.title("Rating Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Rating")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(df["Quantity"],discrete=True, color="orange")
plt.title("Distribution of Quantity Purchase Per Transaction")
plt.xlabel("Quantity")
plt.ylabel("Transaction Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.violinplot(data=df, x="Product line", y="Tax 5%", hue="Product line", legend=False, palette="Set2",)
plt.title("Tax Amount Distribution by product line")
plt.xticks(rotation=45)
plt.xlabel("Product line")
plt.ylabel("Tax 5%")
plt.tight_layout()
plt.show()
```

```python
branch_sales=df.groupby("Branch")["Total"].sum()
plt.figure(figsize=(5,5))
plt.pie(branch_sales, labels=branch_sales.index, autopct="%1.1f%%", colors=["#ff9999","#66b3ff","#99ff99"])
plt.title("Revenue Share by Branch")
plt.tight_layout()
plt.show()
```


###**Insights and findings**###:

- 1.Sales performance varied across product lines , showing that some product categories contributed more to total sales than others.
- 2.Naypytaw recorded the  highest total sales among the three cities, indicating stronger sales performance in that location.
- 3.E-wallet and Cash were the most commonly used payment methods , showing a relatively balanced preference among customers.
- 4.Normal customers were slightly higher than Member customers, indicating a small difference in customer-type distribution.
- 5.Quantity and unit price showed positive relationship with total sales , meaning higher quantities and prices generally contributed to higher sales values.
- 6.Rating for female and male customers showed broadly similar distributions,with no major difference in overall rating patterns.
- 7.A few high-value transaction appeared as potential outliers in the total sales distribution and may require further investigation. 
- 8.The analysis provides usefull insights in to sales perform and customer behaviour.
